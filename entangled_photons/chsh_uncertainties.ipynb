```python
import numpy as np
from uncertainties import ufloat
```

```python
# copy in counts for all polarizer angles and take poissonian errors sqrt(N)
phi_plus = np.array([[1340 , 324 , 352 ,  1512],
                     [1335 , 1390 , 327 , 404],
                     [338 , 1406 , 1300 , 296],
                     [335 , 345 , 1323 , 1387]])
phi_plus = np.array([[ufloat(x, np.sqrt(x)) for x in row] for row in phi_plus])

phi_minus = np.array([[1492 , 355 , 315 ,  1592],
                      [310 , 299 , 1366 , 1479],
                      [256 , 1270 , 1241  , 267],
                      [1340 , 1275 , 271 , 420]])
phi_minus = np.array([[ufloat(x, np.sqrt(x)) for x in row] for row in phi_minus])

hh = np.array([[30 , 27 , 32,  46],
               [507 , 2377 , 2121 ,  301],
               [807 , 4271 , 4191  , 724],
               [310 , 2184 , 2234 , 510]])
hh = np.array([[ufloat(x, np.sqrt(x)) for x in row] for row in hh])


def expect_val(counts, a, b):
    # convert angles to row and column indices
    if a==0: i = 0
    if a==45: i = 1
    if b==22.5: j = 0
    if b==67.5: j = 1

    eval = counts[i,j] - counts[i+2,j] - counts[i,j+2] + counts[i+2,j+2]
    eval /=  counts[i,j] + counts[i+2,j] + counts[i,j+2] + counts[i+2,j+2]
    return eval
```

```python
# compute every expectation value with uncertainty and print in LaTeX table format
angles = [(0, 22.5), (45, 22.5), (0, 67.5), (45, 67.5)]
states = [phi_plus, phi_minus, hh]

for ab in angles:
    print("\t$E(\\qty{"+f"{ab[0]}"+"}{\\degree}, \\qty{"+f"{ab[1]}"+"}{\\degree}$)", end=" ")
    for s in states:
        print(f"&  ${expect_val(s, *ab):S}$", end="  ")
    print("\\\\")
```

```python
# compute S and S_bar with uncertainty
def compute_s_params(counts):
    s = (expect_val(counts, 0, 22.5) + expect_val(counts, 45, 22.5)
         - expect_val(counts, 0, 67.5) + expect_val(counts, 45, 67.5))
    
    s_bar = (expect_val(counts, 0, 22.5) - expect_val(counts, 45, 22.5)
             - expect_val(counts, 0, 67.5) - expect_val(counts, 45, 67.5))
    
    return s, s_bar


print(f"|Phi+> : \t S = {compute_s_params(phi_plus)[0]:S} \t S_bar = {compute_s_params(phi_plus)[1]:S}")
print(f"|Phi-> : \t S = {compute_s_params(phi_minus)[0]:S} \t S_bar = {compute_s_params(phi_minus)[1]:S}")
print(f"|HH>   : \t S = {compute_s_params(hh)[0]:S} \t S_bar = {compute_s_params(hh)[1]:S}")
```
